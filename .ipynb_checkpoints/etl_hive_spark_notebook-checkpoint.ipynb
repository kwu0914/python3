```python
# etl_hive_spark_notebook.ipynb

# --- Setup Spark with Hive support ---
from pyspark.sql import SparkSession
import os

spark = SparkSession.builder \
    .appName("Hive ETL Pipeline") \
    .config("spark.sql.warehouse.dir", "/Volumes/synology-hive/warehouse") \
    .enableHiveSupport() \
    .getOrCreate()
```

```python
# --- Step 1: Load CSV into Spark DataFrame ---
def load_data(csv_name, table_name,csv_path="Volumes/sandisk/datain",db='dc'):
    local_file="/"+csv_path+"/"+csv_name;
    load_path = "file:/"+csv_path+"/"+csv_name;
    full_tb=db+'.'+table_name
    #print(local_file)
    #print(load_path)
    #print(full_tb)
    print("File:"+local_file+" will be loaded into Hive Table dc."+table_name)

    if os.path.isfile(local_file):
        print("File exists - proceed")
        df = spark.read.option("header", True).option("inferSchema", True).csv(load_path)
        print(f"CSV Loaded: {df.count()} rows")
        df.write.mode("overwrite").format("parquet").saveAsTable(full_tb)
        spark.sql("""SELECT * FROM """+full_tb+""" LIMIT 5""").show()
        spark.sql("""select count(*) from """+full_tb+""" """).show()
    else:
        print("File does not exist - Check you path and file name")   
```
